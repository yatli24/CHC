Historical monthly average vs 2015-2024 monthly

```python
# Make a copy of the dataframe to avoid modifying the original
df = eea_ncep.copy()

# Convert 'time' to datetime if it's not already
df['time'] = pd.to_datetime(df['time'])

# Extract year and month
df['year'] = df['time'].dt.year
df['month'] = df['time'].dt.month

# Drop rows with NaN values in the 'precip' column
df_cleaned = df.dropna(subset=['precip'])

# Dictionary for month names
month_names = {1: 'Jan', 2: 'Feb', 3: 'Mar', 4: 'Apr', 5: 'May', 6: 'Jun',
               7: 'Jul', 8: 'Aug', 9: 'Sep', 10: 'Oct', 11: 'Nov', 12: 'Dec'}

# Create subplots grid (5 rows, 2 columns) for 10 years
fig, axes = plt.subplots(5, 2, figsize=(18, 22))
axes = axes.flatten()

for i, year in enumerate(range(2015, 2025)):
    # Calculate monthly average for the given year
    df_year = df_cleaned[df_cleaned['year'] == year]
    monthly_avg = df_year.groupby('month')['precip'].mean().reset_index()
    monthly_avg.columns = ['month', 'avg_precip']

    # Calculate historical monthly average (across all years)
    monthly_historical_avg = df_cleaned.groupby('month')['precip'].mean().reset_index()
    monthly_historical_avg.columns = ['month', 'historical_avg_precip']

    # Merge the two dataframes
    comparison_df = pd.merge(monthly_avg, monthly_historical_avg, on='month')

    # Add month names for better readability
    comparison_df['month_name'] = comparison_df['month'].map(month_names)

    # Sort by month for proper plotting
    comparison_df = comparison_df.sort_values('month')

    # Plot on the corresponding subplot
    ax = axes[i]
    ax.plot(comparison_df['month_name'], comparison_df['historical_avg_precip'], 'o-', linewidth=2, markersize=8, label='Historical Average')
    ax.plot(comparison_df['month_name'], comparison_df['avg_precip'], 'o-', linewidth=2, markersize=8, label=f'{year} Average')

    ax.set_xlabel('Month', fontsize=10)
    ax.set_ylabel('Precipitation', fontsize=10)
    ax.set_title(f'{year} Monthly Precipitation vs Historical Average', fontsize=12)
    ax.grid(True, linestyle='--', alpha=0.7)
    ax.legend(fontsize=8)

    # Rotate month names for better readability
    ax.set_xticklabels(comparison_df['month_name'], rotation=45)

# Hide any unused subplots (if applicable)
for i in range(len(range(2015, 2025)), len(axes)):
    fig.delaxes(axes[i])

# Adjust layout for better spacing
plt.tight_layout()
plt.show()
```

2024 with all lead times.

```python
# Multiply predicted_precip by 30
df['predicted_precip'] = df['predicted_precip'] * 30

# Define lead times to analyze
lead_times = [0.5, 1.5, 2.5, 3.5, 4.5, 5.5, 6.5, 7.5, 8.5, 9.5]

# Define month names for plotting
month_names = {1: 'Jan', 2: 'Feb', 3: 'Mar', 4: 'Apr', 5: 'May', 6: 'Jun',
               7: 'Jul', 8: 'Aug', 9: 'Sep', 10: 'Oct', 11: 'Nov', 12: 'Dec'}

# Create a figure with subplots - 4 rows, 3 columns for 10 lead times
fig, axes = plt.subplots(4, 3, figsize=(18, 16))
axes = axes.flatten()  # Flatten to make indexing easier

# Store all results for final table
all_results = []

# Loop through each lead time and create a subplot
for i, lead_time in enumerate(lead_times):
    # Filter for the specific lead time
    df_lead = df[df['lead_time'] == lead_time]

    # Drop rows with NaN values in the predicted_precip column
    df_cleaned = df_lead.dropna(subset=['predicted_precip'])

    # Calculate monthly average for 2024 with this lead time
    df_2024 = df_cleaned[df_cleaned['year'] == 2024]
    monthly_avg_2024 = df_2024.groupby('month')['predicted_precip'].mean().reset_index()
    monthly_avg_2024.columns = ['month', 'avg_precip_2024']

    # Calculate historical monthly average for all years with this lead time
    monthly_historical_avg = df_cleaned.groupby('month')['predicted_precip'].mean().reset_index()
    monthly_historical_avg.columns = ['month', 'historical_avg_precip']

    # Merge the two dataframes
    comparison_df = pd.merge(monthly_avg_2024, monthly_historical_avg, on='month', how='outer')

    # Add month names
    comparison_df['month_name'] = comparison_df['month'].map(month_names)

    # Sort by month for proper plotting
    comparison_df = comparison_df.sort_values('month')

    # Store results for the table
    for _, row in comparison_df.iterrows():
        all_results.append({
            'lead_time': lead_time,
            'month': row['month'],
            'month_name': row['month_name'],
            'avg_2024': row['avg_precip_2024'],
            'historical_avg': row['historical_avg_precip']
        })

    # Plot on the respective subplot
    ax = axes[i]
    ax.plot(comparison_df['month_name'], comparison_df['historical_avg_precip'], 'o-', linewidth=2, markersize=6, label='Historical Avg')
    ax.plot(comparison_df['month_name'], comparison_df['avg_precip_2024'], 'o-', linewidth=2, markersize=6, label='2024 Avg')

    ax.set_title(f'Lead Time {lead_time}', fontsize=12)
    ax.set_xlabel('Month', fontsize=10)
    ax.set_ylabel('Predicted Precipitation (×30)', fontsize=10)
    ax.grid(True, linestyle='--', alpha=0.7)
    ax.tick_params(axis='x', rotation=45)

    # Add legend to only the first subplot to avoid redundancy
    if i == 0:
        ax.legend(fontsize=10)

# Set the main title
fig.suptitle('2024 vs Historical Monthly Predicted Precipitation Across Lead Times)', fontsize=16)

# Remove any unused subplots
for j in range(i+1, len(axes)):
    fig.delaxes(axes[j])

# Adjust layout
plt.tight_layout()
plt.subplots_adjust(top=0.95)

plt.show()

# Create a consolidated dataframe from results
results_df = pd.DataFrame(all_results)

# Create a pivot table for easier reading
pivot_df = results_df.pivot_table(
    index=['lead_time', 'month', 'month_name'],
    values=['avg_2024', 'historical_avg'],
    aggfunc='first'
).reset_index()

# Sort by lead time and month
pivot_df = pivot_df.sort_values(['lead_time', 'month'])

# Print a sample of the results
print("Predicted Precipitation Comparison by Lead Time and Month (2024 vs Historical):")
print(pivot_df.head(10).round(3))
```

MISSING DATA FUNCTION

```python
def find_missing_date_ranges(directory):
    results = []

    # Loop over files in the directory
    for file in os.listdir(directory):
        if file.endswith("_merged.nc"):  # Ensure only relevant files are processed
            file_path = os.path.join(directory, file)

            try:
                # Open the netCDF file
                ds = xr.open_dataset(file_path)

                # Extract time variable and convert to pandas datetime
                if 'time' in ds.variables:
                    time_data = pd.to_datetime(ds['time'].values)

                    # Identify gaps in the time series
                    full_range = pd.date_range(start=time_data.min(), end=time_data.max(), freq='ME')
                    missing_dates = full_range.difference(time_data)

                    if not missing_dates.empty:
                        # Identify missing data ranges
                        missing_periods = []
                        start_date = missing_dates[0]

                        for i in range(1, len(missing_dates)):
                            if (missing_dates[i] - missing_dates[i-1]).days > 31:  # Check for break in sequence
                                end_date = missing_dates[i-1]
                                missing_periods.append(f"{start_date.strftime('%B %Y')} to {end_date.strftime('%B %Y')}")
                                start_date = missing_dates[i]

                        # Append last missing range
                        missing_periods.append(f"{start_date.strftime('%B %Y')} to {missing_dates[-1].strftime('%B %Y')}")

                        # Format file name
                        file_name = file.replace("_merged.nc", "")
                        results.append((file_name, ", ".join(missing_periods)))

                ds.close()
            except Exception as e:
                print(f"Error processing {file}: {e}")

    # Sort results by model name (last part of the file name)
    results.sort(key=lambda x: x[0].split("_")[-1])

    return results

# Example usage
directory = "data/netCDF/"
missing_data_info = find_missing_date_ranges(directory)

# Print the results
for file_name, missing_range in missing_data_info:
    print(f"{file_name}: {missing_range}")
```

```python
# Define file paths for all models
model_paths = {
    "NCEP": "data/netCDF/eastern_east_africa_NCEP_merged.nc",
    "NASA": "data/netCDF/eastern_east_africa_NASA_merged.nc",
    "CanESM5": "data/netCDF/eastern_east_africa_CanESM5_merged.nc",
    "CCSM4": "data/netCDF/eastern_east_africa_CCSM4_merged.nc",
    "GFDL": "data/netCDF/eastern_east_africa_GFDL_merged.nc"
}


# Dictionary to store results for each model
model_results = {}

for model, path in model_paths.items():
    try:
        # Load dataset
        dataset = xr.open_dataset(path)

        ##### when slicing to the precip or predicted_precip, then one of the columns is gone, that clears out the whole column leading to missing data. How do I make it possible to slice the data into dataframe and not miss out on any of the columns?
        # Convert to DataFrame
        df = dataset.to_dataframe().reset_index()
        df["predicted_precip"] = df["predicted_precip"] * 30

        # Ensure the 'time' column is in datetime format
        df['time'] = pd.to_datetime(df['time'])

        # Extract year and month
        df['year'] = df['time'].dt.year
        df['month'] = df['time'].dt.month

        # Compute historical monthly averages (all years)
        historical_avg = df.groupby('month')['precip'].mean()

        # Compute yearly monthly averages for 2020-2024
        yearly_avg = df[(df['year'] >= 2020) & (df['year'] <= 2024)].groupby(['year', 'month'])['precip'].mean().unstack(level=0)

        # Store results
        model_results[model] = {"historical_avg": historical_avg, "yearly_avg": yearly_avg}

    except FileNotFoundError:
        print(f"File not found: {path}, skipping {model}.")
    except Exception as e:
        print(f"Error processing {model}: {e}")

# Plot separate figures for each model
for model, results in model_results.items():
    plt.figure(figsize=(10, 5))

    historical_avg = results["historical_avg"]
    yearly_avg = results["yearly_avg"]

    # Plot historical average
    plt.plot(historical_avg.index, historical_avg.values, marker='o', linestyle='-', linewidth=2, label=f"{model} Historical Avg")

    # Plot yearly averages from 2020 to 2024
    for year in yearly_avg.columns:
        plt.plot(yearly_avg.index, yearly_avg[year], marker='s', linestyle='--', alpha=0.7, label=f"{model} {year}")

    plt.xlabel("Month")
    plt.ylabel("Average Precipitation")
    plt.title(f"{model} - Monthly Average Precipitation (2020-2024)")
    plt.xticks(range(1, 13), ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])
    plt.legend(loc='upper right')
    plt.grid(True)

    # Show the plot
    plt.show()
```

Overall Seasonal Pattern

Notice the peaks and valleys in the historical average line (green). For many parts of Eastern Africa, there is a primary rainy season around March–May and often a secondary rainy season around October–December.
Compare the individual years (2020–2024) to see how they follow or deviate from this typical seasonal pattern.
Comparison With the Historical Average

Identify months where a particular year’s precipitation is above or below the historical average. For instance:
If the orange line (2020) is above the green line in April, it suggests April 2020 was wetter than usual.
If the red line (2022) is below the green line in November, it suggests November 2022 was drier than the long-term norm.
This helps you see how unusual or typical each year’s precipitation was in any given month.
Interannual Variability

Look for differences among the years themselves. Do certain years consistently trend higher or lower than others?
For instance, if 2021 consistently sits above the historical average across multiple months, it might indicate a wetter-than-normal year overall, whereas 2023 might be close to or below average in most months, indicating drier conditions.


```python
# Dictionary of dictionaries:
# Outer keys = region names
# Inner keys = model names
# Values = file paths to the corresponding NetCDF files

region_model_paths = {
    "eastern_east_africa": {
    "NCEP": "data/netCDF/eastern_east_africa_NCEP_merged.nc",
    "NASA": "data/netCDF/eastern_east_africa_NASA_merged.nc",
    "CanESM5": "data/netCDF/eastern_east_africa_CanESM5_merged.nc",
    "CCSM4": "data/netCDF/eastern_east_africa_CCSM4_merged.nc",
    "GFDL": "data/netCDF/eastern_east_africa_GFDL_merged.nc"
    },
    "south_sudan": {
        "NCEP": "data/netCDF/south_sudan_NCEP_merged.nc",
        "NASA": "data/netCDF/south_sudan_NASA_merged.nc",
        "CanESM5": "data/netCDF/south_sudan_CanESM5_merged.nc",
        "CCSM4": "data/netCDF/south_sudan_CCSM4_merged.nc",
        "GFDL": "data/netCDF/south_sudan_GFDL_merged.nc"
    },
    "eastern_ukraine": {
        "NCEP": "data/netCDF/eastern_ukraine_NCEP_merged.nc",
        "NASA": "data/netCDF/eastern_ukraine_NASA_merged.nc",
        "CanESM5": "data/netCDF/eastern_ukraine_CanESM5_merged.nc",
        "CCSM4": "data/netCDF/eastern_ukraine_CCSM4_merged.nc",
        "GFDL": "data/netCDF/eastern_ukraine_GFDL_merged.nc"
    },
    "lake_victoria_basin": {
        "NCEP": "data/netCDF/lake_victoria_basin_NCEP_merged.nc",
        "NASA": "data/netCDF/lake_victoria_basin_NASA_merged.nc",
        "CanESM5": "data/netCDF/lake_victoria_basin_CanESM5_merged.nc",
        "CCSM4": "data/netCDF/lake_victoria_basin_CCSM4_merged.nc",
        "GFDL": "data/netCDF/lake_victoria_basin_GFDL_merged.nc"
    }
}

# Loop through each region
for region, model_paths in region_model_paths.items():
    print(f"=== Processing region: {region} ===")

    # Dictionary to store results for each model within this region
    model_results = {}

    # Loop through each model for the current region
    for model, path in model_paths.items():
        try:
            # Load dataset
            dataset = xr.open_dataset(path)

            # Convert to DataFrame
            df = dataset.to_dataframe().reset_index()
            df["predicted_precip"] = df["predicted_precip"] * 30

            # Ensure the 'time' column is in datetime format
            df['time'] = pd.to_datetime(df['time'])

            # Extract year and month
            df['year'] = df['time'].dt.year
            df['month'] = df['time'].dt.month

            # Compute historical monthly averages (all years in the dataset)
            historical_avg = df.groupby('month')['precip'].mean()

            # Compute yearly monthly averages for 2020-2024
            yearly_avg = (
                df[(df['year'] >= 2020) & (df['year'] <= 2024)]
                .groupby(['year', 'month'])['precip']
                .mean()
                .unstack(level=0)
            )

            # Store results
            model_results[model] = {"historical_avg": historical_avg, "yearly_avg": yearly_avg}

        except FileNotFoundError:
            print(f"File not found: {path}, skipping {model}.")
        except Exception as e:
            print(f"Error processing {model}: {e}")

    # Now plot separate figures for each model within this region
    for model, results in model_results.items():
        plt.figure(figsize=(10, 5))

        historical_avg = results["historical_avg"]
        yearly_avg = results["yearly_avg"]

        # Plot historical average
        plt.plot(
            historical_avg.index,
            historical_avg.values,
            marker='o',
            linestyle='-',
            linewidth=2,
            label=f"{model} Historical Avg"
        )

        # Plot yearly averages from 2020 to 2024
        if yearly_avg is not None:
            for year in yearly_avg.columns:
                plt.plot(
                    yearly_avg.index,
                    yearly_avg[year],
                    marker='s',
                    linestyle='--',
                    alpha=0.7,
                    label=f"{model} {year}"
                )

        plt.xlabel("Month")
        plt.ylabel("Average Precipitation")
        plt.title(f"{region.upper()} - {model} Monthly Avg Precipitation (2020-2024)")
        plt.xticks(range(1, 13), ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])
        plt.legend(loc='upper right')
        plt.grid(True)
        plt.show()
```

```python


# Define file paths for all regions and models
region_model_paths = {
    "eastern_east_africa": {
        "NCEP": "data/netCDF/eastern_east_africa_NCEP_merged.nc",
        "NASA": "data/netCDF/eastern_east_africa_NASA_merged.nc",
        "CanESM5": "data/netCDF/eastern_east_africa_CanESM5_merged.nc",
        "CCSM4": "data/netCDF/eastern_east_africa_CCSM4_merged.nc",
        "GFDL": "data/netCDF/eastern_east_africa_GFDL_merged.nc"
    },
    "south_sudan": {
        "NCEP": "data/netCDF/south_sudan_NCEP_merged.nc",
        "NASA": "data/netCDF/south_sudan_NASA_merged.nc",
        "CanESM5": "data/netCDF/south_sudan_CanESM5_merged.nc",
        "CCSM4": "data/netCDF/south_sudan_CCSM4_merged.nc",
        "GFDL": "data/netCDF/south_sudan_GFDL_merged.nc"
    },
    "eastern_ukraine": {
        "NCEP": "data/netCDF/eastern_ukraine_NCEP_merged.nc",
      "NASA": "data/netCDF/eastern_ukraine_NASA_merged.nc",
        "CanESM5": "data/netCDF/eastern_ukraine_CanESM5_merged.nc",
        "CCSM4": "data/netCDF/eastern_ukraine_CCSM4_merged.nc",
        "GFDL": "data/netCDF/eastern_ukraine_GFDL_merged.nc"
    },
    "lake_victoria_basin": {
        "NCEP": "data/netCDF/lake_victoria_basin_NCEP_merged.nc",
        "NASA": "data/netCDF/lake_victoria_basin_NASA_merged.nc",
        "CanESM5": "data/netCDF/lake_victoria_basin_CanESM5_merged.nc",
        "CCSM4": "data/netCDF/lake_victoria_basin_CCSM4_merged.nc",
        "GFDL": "data/netCDF/lake_victoria_basin_GFDL_merged.nc"
    }
}

# Dictionary to hold aggregated data for each region
region_data = {}

# Loop over each region, load and combine data from all models
for region, models in region_model_paths.items():
    df_list = []
    for model, path in models.items():
        try:
            # Load dataset
            ds = xr.open_dataset(path)
            # Convert entire dataset to DataFrame (to keep all variables)
            df = ds.to_dataframe().reset_index()
            df["predicted_precip"] = df["predicted_precip"] * 30
            # Convert time to datetime and extract year/month
            df['time'] = pd.to_datetime(df['time'])
            df['year'] = df['time'].dt.year
            df['month'] = df['time'].dt.month
            df_list.append(df)
        except Exception as e:
            print(f"Error processing {region} {model}: {e}")
    if df_list:
        # Combine data from all models for the region
        region_data[region] = pd.concat(df_list, axis=0)

# Define years to plot (rows)
years_to_plot = [2020, 2021, 2022, 2023, 2024]
# Define regions (columns) in a specific order
regions = list(region_model_paths.keys())

# Create grid: 5 rows (years) x 4 columns (regions)
fig, axes = plt.subplots(nrows=5, ncols=4, figsize=(20, 20), sharex=True, sharey=True)

# Loop over each region (column) and year (row)
for col_idx, region in enumerate(regions):
    # Get the combined dataframe for this region
    df_region = region_data.get(region)
    if df_region is None:
        continue
    # Compute the historical monthly average for predicted_precip across all years
    historical_avg = df_region.groupby('month')["precip"].mean()

    for row_idx, year in enumerate(years_to_plot):
        ax = axes[row_idx, col_idx]
        # Filter data for the current year
        df_year = df_region[df_region['year'] == year]
        # Compute monthly average for this year
        year_avg = df_year.groupby('month')["precip"].mean()

        # Plot historical average (common for region)
        ax.plot(historical_avg.index, historical_avg.values,
                marker='o', linestyle='-', linewidth=2, color='gray', label="Hist. Avg")
        # Plot the current year's monthly averages
        ax.plot(year_avg.index, year_avg.values,
                marker='s', linestyle='--', alpha=0.8, label=str(year))

        ax.set_xticks(range(1, 13))
        ax.set_xticklabels(["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"], rotation=45)
        ax.grid(True)
        if row_idx == 0:
            ax.set_title(region.replace("_", " ").title())
        if col_idx == 0:
            ax.set_ylabel(str(year))
        else:
            ax.set_ylabel("")
        # Optionally, add a legend to the top row
        if row_idx == 0 and col_idx == 0:
            ax.legend(fontsize=8)

# Set common x and y labels for the entire grid
fig.text(0.5, 0.04, 'Month', ha='center', fontsize=14)
fig.text(0.04, 0.5, 'Predicted Precipitation', va='center', rotation='vertical', fontsize=14)

plt.tight_layout(rect=[0.05, 0.05, 1, 0.95])
plt.suptitle("Historical vs. Yearly Monthly Average Predicted Precipitation\n(Columns: Regions, Rows: Years 2020-2024)", fontsize=16)
plt.show()
```
