```python
# xarray to read NETCDF
import numpy as np
import pandas as pd
import dask.dataframe as dd
import xarray as xr
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Opening CHIRPS data
chirps = xr.open_dataset('data/CHIRPS/chirps-v2.0.monthly.nc')
```

```python
# Subset chirps to South Sudan and convert to Dataframe
south_sudan = chirps.sel(latitude=slice(3.5, 12.5), longitude=slice(25, 35))
south_sudan_df = south_sudan.to_dataframe()
# Calculate region average
south_sudan_region_average = south_sudan_df.groupby(['time']).mean().reset_index()
south_sudan_region_average['month'] = south_sudan_region_average['time'].dt.month
south_sudan_monthly_average = south_sudan_region_average.drop('time', axis=1).groupby('month').mean()
south_sudan_monthly_average
```
